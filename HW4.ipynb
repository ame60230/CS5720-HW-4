```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")

sentence = "NLP techniques are used in virtual assistants like Alexa and Siri."

def preprocess_text(sentence):
    tokens = word_tokenize(sentence)

    stop_words = set(stopwords.words("english"))

    filtered_tokens = [
        word for word in tokens
        if word.lower() not in stop_words and word.isalpha()
    ]

    stemmer = PorterStemmer()

    stemmed_words = [
        stemmer.stem(word)
        for word in filtered_tokens
    ]

    print("Original Tokens:")
    print(tokens)

    print("\nTokens Without Stopwords:")
    print(filtered_tokens)

    print("\nStemmed Words:")
    print(stemmed_words)

preprocess_text(sentence)
```

1. Stemming cuts a word down to a basic root form, while lemmatization tries to return a real dictionary form of the word. For example, stemming "running" may give "run", and lemmatization can also give "run" when it knows the word is being used as a verb.

2. Removing stop words can be useful because common words like "the" and "in" may not add much meaning and can make the data larger. It can be harmful when those words actually affect the meaning of a sentence, especially in tasks like sentiment analysis or understanding context.

```python
import numpy as np

Q = np.array([
    [1, 0, 1, 0],
    [0, 1, 0, 1]
], dtype=float)

K = np.array([
    [1, 0, 1, 0],
    [0, 1, 0, 1]
], dtype=float)

V = np.array([
    [1, 2, 3, 4],
    [5, 6, 7, 8]
], dtype=float)

def scaled_dot_product_attention(Q, K, V):
    scores = np.dot(Q, K.T)

    d = K.shape[1]
    scaled_scores = scores / np.sqrt(d)

    exp_scores = np.exp(
        scaled_scores - np.max(scaled_scores, axis=1, keepdims=True)
    )

    attention_weights = exp_scores / np.sum(
        exp_scores,
        axis=1,
        keepdims=True
    )

    output = np.dot(attention_weights, V)

    return attention_weights, output


attention_weights, output = scaled_dot_product_attention(Q, K, V)

print("Attention Weights:")
print(np.round(attention_weights, 3))

print("\nFinal Output:")
print(np.round(output, 3))
```

1. We divide by the square root of d so the attention scores do not get too large. This helps keep the softmax values more balanced and makes training more stable.

2. Self-attention lets each word look at the other words in the sentence and figure out which ones are important. This helps the model understand the context and relationships between words.
